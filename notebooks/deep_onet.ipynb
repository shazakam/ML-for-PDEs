```python
import torch
import sys
sys.path.append('..')
from models.forecasting.deep_onet import DeepONet
from datasets.heat_dataset import HeatONetDataset
import yaml
from torch.utils.data import DataLoader, random_split
with open('../configs/training_configs/deep_onet_heat_training.yaml') as f:
    config = yaml.safe_load(f)
```

```python
train_data = HeatONetDataset('../'+config['training_data_path'], config['field_keys'])
train_dl = DataLoader(train_data, batch_size= 16)
```

```python
for batch in train_dl:
    k = batch
    break
```

```python
model = DeepONet(
    conv_branch_layers=config['conv_branch_layers'],
    conv_branch_activations=config['conv_branch_activations'],
    stride_branch=config['stride_branch'],
    ffn_branch_layers=config['ffn_branch_layers'],
    ffn_branch_activations=config['ffn_branch_activations'],
    ffn_trunk_layers=config['ffn_trunk_layers'],
    ffn_trunk_activations=config['ffn_trunk_activations'],
    dropout=config['dropout'],
    optimiser=config['optimiser'],
    learning_rate=config['learning_rate'],
)
```

```python
model.eval()
with torch.no_grad():
    out = model(k[0], k[1])
```
